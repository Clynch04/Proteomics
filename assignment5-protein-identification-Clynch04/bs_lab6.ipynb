```python
# Monoisotopic residue masses
aa_masses = {
    'A': 71.03711,  'R': 156.10111, 'N': 114.04293, 'D': 115.02694,
    'C': 103.00919, 'E': 129.04259, 'Q': 128.05858, 'G': 57.02146,
    'H': 137.05891, 'I': 113.08406, 'L': 113.08406, 'K': 128.09496,
    'M': 131.04049, 'F': 147.06841, 'P': 97.05276,  'S': 87.03203,
    'T': 101.04768, 'W': 186.07931, 'Y': 163.06333, 'V': 99.06841
}
# Mass of a proton and water
PROTON = 1.0073
H2O = 18.01056
```

```python
def calculate_by_ions(peptide):
    # Calculate b-ions and y-ions for a given peptide sequence
    b_ions = []
    y_ions = []

    # b-ions
    running_mass = 0
    for i in range(len(peptide) - 1): # Exclude the last amino acid for b-ions
        fragment = peptide[:i+1] # Get the b-ion fragment
        running_mass += aa_masses[peptide[i]] # Add the mass of the current amino acid
        b_mass = running_mass + PROTON # Add the mass of a proton for b-ions
        
        b_ions.append({
            "ion": f"b{i+1}",
            "sequence": fragment,
            "m/z": round(b_mass, 4)
        })

    # y-ions
    running_mass = H2O  # Start with water mass for y-ions
    for i in range(len(peptide)-1, 0, -1): # Start from the last amino acid and move backwards
        fragment = peptide[i:]
        running_mass += aa_masses[peptide[i]]
        y_mass = running_mass + PROTON
        
        y_ions.append({
            "ion": f"y{len(peptide)-i}",
            "sequence": fragment,
            "m/z": round(y_mass, 4)
        })

    return b_ions, y_ions
```

```python
selected_index = 0  # Change this index to select a different peptide
peptide = df['Sequence'][selected_index]

print(f"Calculating b- and y-ions for peptide: {peptide}")
b_ions, y_ions = calculate_by_ions(peptide) # Calculate b- and y-ions for the selected peptide

print("b-ions:")
for ion in b_ions:
    print(f"{ion['ion']}  {ion['sequence']}  {ion['m/z']}")

print("\ny-ions:")
for ion in y_ions:
    print(f"{ion['ion']}  {ion['sequence']}  {ion['m/z']}")
```

```python
import pymzml
import numpy as np
import holoviews as hv
hv.extension('bokeh')

mzml_file = df['Raw data location'][selected_index]

# I changed the path to the lab 6 directory after running the postprocessing code
mzml_file = mzml_file.replace('/home/cplcgns/lab6/mzML_files/', '/home/cplcgns/proteomics/lab6/mzML_files/')

# The original file paths in the CSV point to .mgf files, but we need .mzML files for pymzml
mzml_file = mzml_file.replace('.mgf', '.mzML')

print(f"Loading mzML file: {mzml_file}")

run = pymzml.run.Reader(mzml_file) # Load the mzML file using pymzml
spec = run[df['Spectrum ID'][selected_index]] # Access the spectrum using the Spectrum ID from the DataFrame
```

```python
# Code block taken from lab 5 sample code to plot the spectrum

print("Spectrum MS level:", spec.ms_level)

mz_list = []
intensities_list = []
for mz, i in spec.peaks("centroided"):
    mz_list.append(mz)
    intensities_list.append(i)

df_spec = pd.DataFrame({
    "mz": mz_list,
    "intensity": intensities_list
})

ms_spikes = hv.Spikes(
    df_spec, 
    kdims="mz", 
    vdims="intensity"
).opts(
    width=900,
    height=350,
    line_width=1.0,
    color='black',
    tools=['box_zoom', 'hover', 'pan', 'wheel_zoom', 'reset', 'save'],
    active_tools=['box_zoom'],
    ylim=(0, None),
    xlabel="m/z",
    ylabel="Intensity",
    title="Annotated MS/MS Spectrum"
)

ms_spikes
```

```python
def match_ions_haspeak(theoretical_ions, spec, ppm_tol=10):
    # Match theoretical ions to observed peaks in the spectrum using the has_peak method
    matches = []

    for ion in theoretical_ions:
        theo_mz = ion["m/z"] # Get the theoretical m/z value for the ion
        peak_list = spec.has_peak(theo_mz) # Use has_peak to find peaks matching the theoretical m/z within the specified tolerance (0.1 Da by default)

        if peak_list:
            observed_mz, intensity = peak_list[0] # Get the observed m/z and intensity of the matching peak

            ppm_error = (observed_mz - theo_mz) / theo_mz * 1e6 # Calculate the ppm error between the observed m/z and the theoretical m/z

            if abs(ppm_error) <= ppm_tol: # Only consider matches within the specified ppm tolerance
                matches.append({
                    "ion": ion["ion"],
                    "observed_mz": observed_mz,
                    "intensity": intensity
                })

    return matches
```

```python
def highlight_matches(ms_spikes, matches, ppm=10):
    # Create an overlay of the original spectrum and the matched peaks highlighted with triangles
    overlays = [ms_spikes]
    label_data = []

    for m in matches:

        mz = m["observed_mz"]
        intensity = m["intensity"]

        delta = (ppm * mz) / 1e6 # Calculate the m/z range for the triangle based on the ppm tolerance

        triangle_df = pd.DataFrame({
            "mz": [mz - delta, mz, mz + delta], # Create a small triangle around the observed m/z value
            "intensity": [0, intensity, 0]
        })

        if m["ion"].startswith("b"): # Color b-ions blue and y-ions red
            color = "blue"
        else:
            color = "red"

        # Create an area plot for the triangle to highlight the matched peak
        peak_triangle = hv.Area(
            triangle_df,
            kdims="mz",
            vdims="intensity"
        ).opts(
            color=color,
            alpha=0.6,
            line_color=color,
            line_width=1
        )

        # Add the triangle to the overlays and prepare label data for the ion annotation
        overlays.append(peak_triangle)
        label_data.append((mz, intensity, m["ion"]))

    labels = hv.Labels(label_data).opts(
        text_font_size="8pt"
    )

    overlays.append(labels)

    return hv.Overlay(overlays)
```

```python
# Combine b-ions and y-ions into a single list of theoretical ions for matching
theoretical_ions = b_ions + y_ions

# Match the theoretical ions to the observed spectrum using the has_peak method with a ppm tolerance of 10 ppm
matches = match_ions_haspeak(theoretical_ions, spec, ppm_tol=10)

# Create an annotated plot of the spectrum with matched peaks highlighted
annotated_plot = highlight_matches(ms_spikes, matches, ppm=10)

annotated_plot
```
