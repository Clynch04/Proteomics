```python
search_word = "Phospho"
# Use str.contains to filter rows where 'Modifications' contains the search word (Phospho)
result = df[df['Modifications'].str.contains(rf'\b{search_word}\b', case=False, na=False)]

print("Rows containing the modification 'Phospho':")
print(result[["Spectrum ID", "Sequence", "Modifications"]])
```

```python
# Monoisotopic residue masses
aa_masses = {
    'A': 71.03711,  'R': 156.10111, 'N': 114.04293, 'D': 115.02694,
    'C': 103.00919, 'E': 129.04259, 'Q': 128.05858, 'G': 57.02146,
    'H': 137.05891, 'I': 113.08406, 'L': 113.08406, 'K': 128.09496,
    'M': 131.04049, 'F': 147.06841, 'P': 97.05276,  'S': 87.03203,
    'T': 101.04768, 'W': 186.07931, 'Y': 163.06333, 'V': 99.06841
}
# Mass of proton, water, and phosphote group
PROTON = 1.0073
H2O = 18.01056
PHOSPHO = 79.96633
```

```python
def parse_phospho_site(mod_string):
    # This function extracts the phospho site from the modification string
    if isinstance(mod_string, str) and "Phospho" in mod_string:
        site = int(mod_string.split(":")[1].strip())
        return site
    return None
```

```python
def calculate_by_ions_with_phospho(peptide, mod_string=None):
    # This function calculates b- and y-ions for a given peptide sequence, accounting for phosphorylation modifications
    b_ions = []
    y_ions = []

    # Parse the phospho site from the modification string
    phospho_site = parse_phospho_site(mod_string)

    # Convert to 0-based index if present
    if phospho_site:
        phospho_site -= 1

    # b-ions
    running_mass = 0
    for i in range(len(peptide) - 1):
        # Add the mass of the current amino acid
        running_mass += aa_masses[peptide[i]]

        # Add phosphorylation if this fragment contains modified residue
        if phospho_site is not None and i >= phospho_site:
            fragment_mass = running_mass + PHOSPHO
            contains_phospho = True
        else:
            fragment_mass = running_mass
            contains_phospho = False

        # Calculate m/z for the b-ion (charge +1)
        b_mass = fragment_mass + PROTON

        b_ions.append({
            "ion": f"b{i+1}",
            "sequence": peptide[:i+1],
            "m/z": b_mass,
            "contains_phospho": contains_phospho
        })

    # y-ions
    running_mass = H2O
    for i in range(len(peptide)-1, 0, -1): # Start from the last amino acid and move backwards

        running_mass += aa_masses[peptide[i]]

        # Check if modified residue is in this C-terminal fragment
        if phospho_site is not None and i <= phospho_site:
            fragment_mass = running_mass + PHOSPHO
            contains_phospho = True
        else:
            fragment_mass = running_mass
            contains_phospho = False

        y_mass = fragment_mass + PROTON

        y_ions.append({
            "ion": f"y{len(peptide)-i}",
            "sequence": peptide[i:],
            "m/z": y_mass,
            "contains_phospho": contains_phospho
        })

    return b_ions, y_ions
```

```python
selected_index = 19963  # Change this index to select a different peptide
mod_string = df['Modifications'][selected_index] # Get the modification string for the selected peptide
peptide = df['Sequence'][selected_index]

print(f"Calculating b- and y-ions for peptide: {peptide}")
b_ions, y_ions = calculate_by_ions_with_phospho(peptide, mod_string) # Calculate b- and y-ions for the selected peptide

print("b-ions:")
for ion in b_ions:
    print(f"{ion['ion']}  {ion['sequence']}  {ion['m/z']}")

print("\ny-ions:")
for ion in y_ions:
    print(f"{ion['ion']}  {ion['sequence']}  {ion['m/z']}")
```

```python
import pymzml
import numpy as np
import holoviews as hv
hv.extension('bokeh')

mzml_file = df['Raw data location'][selected_index]

# The original file paths in the CSV point to .mgf files, but we need .mzML files for pymzml
mzml_file = mzml_file.replace('.mgf', '.mzML')

print(f"Loading mzML file: {mzml_file}")

run = pymzml.run.Reader(mzml_file) # Load the mzML file using pymzml
spec = run[df['Spectrum ID'][selected_index]] # Access the spectrum using the Spectrum ID from the DataFrame
```

```python
# Code block taken from lab 5 sample code to plot the spectrum

print("Spectrum MS level:", spec.ms_level)

mz_list = []
intensities_list = []
for mz, i in spec.peaks("centroided"):
    mz_list.append(mz)
    intensities_list.append(i)

df_spec = pd.DataFrame({
    "mz": mz_list,
    "intensity": intensities_list
})

ms_spikes = hv.Spikes(
    df_spec, 
    kdims="mz", 
    vdims="intensity"
).opts(
    width=900,
    height=350,
    line_width=1.0,
    color='black',
    tools=['box_zoom', 'hover', 'pan', 'wheel_zoom', 'reset', 'save'],
    active_tools=['box_zoom'],
    ylim=(0, None),
    xlabel="m/z",
    ylabel="Intensity",
    title="Annotated MS/MS Spectrum"
)

ms_spikes
```

```python
def match_ions_haspeak(theoretical_ions, spec, ppm_tol=10):
    # Match theoretical ions to observed peaks in the spectrum using the has_peak method
    matches = []

    for ion in theoretical_ions:
        theo_mz = ion["m/z"] # Get the theoretical m/z value for the ion
        peak_list = spec.has_peak(theo_mz) # Use has_peak to find peaks matching the theoretical m/z within the specified tolerance (0.1 Da by default)

        if peak_list:
            observed_mz, intensity = peak_list[0] # Get the observed m/z and intensity of the matching peak

            ppm_error = (observed_mz - theo_mz) / theo_mz * 1e6 # Calculate the ppm error between the observed m/z and the theoretical m/z

            if abs(ppm_error) <= ppm_tol: # Only consider matches within the specified ppm tolerance
                matches.append({
                    "ion": ion["ion"],
                    "observed_mz": observed_mz,
                    "intensity": intensity,
                    "contains_phospho": ion["contains_phospho"]
                })

    return matches
```

```python
def highlight_matches(ms_spikes, matches, ppm=10):
    # Create an overlay of the original spectrum and the matched peaks highlighted with triangles
    overlays = [ms_spikes]
    label_data = []

    for m in matches:

        mz = m["observed_mz"]
        intensity = m["intensity"]

        delta = (ppm * mz) / 1e6 # Calculate the m/z range for the triangle based on the ppm tolerance

        triangle_df = pd.DataFrame({
            "mz": [mz - delta, mz, mz + delta], # Create a small triangle around the observed m/z value
            "intensity": [0, intensity, 0]
        })

        is_phospho = m["contains_phospho"]

        # Set colors based on ion type and whether it contains a phospho modification
        if m["ion"].startswith("b"):
            base_color = "blue"
            phospho_color = "cyan"
        else:
            base_color = "red"
            phospho_color = "magenta"

        # Choose color based on whether the ion contains a phospho modification
        color = phospho_color if is_phospho else base_color

        # Create an area plot for the triangle to highlight the matched peak
        peak_triangle = hv.Area(
            triangle_df,
            kdims="mz",
            vdims="intensity"
        ).opts(
            # Color scheme for matched peaks, with higher opacity and thicker lines for phospho-containing ions
            color=color,
            alpha=0.85 if is_phospho else 0.6,
            line_color=color,
            line_width=3 if is_phospho else 1
        )

        # Add the triangle to the overlays and prepare label data for the ion annotation
        overlays.append(peak_triangle)
        label_text = m["ion"] + "*" if is_phospho else m["ion"]
        label_data.append((
            mz,
            intensity * (1.12 if is_phospho else 1.05),
            label_text,
            is_phospho
        ))

    # Split labels so we can style separately
    normal_labels = [(x,y,t) for x,y,t,p in label_data if not p]
    phospho_labels = [(x,y,t) for x,y,t,p in label_data if p]

    overlays.append(
        hv.Labels(normal_labels).opts(
            text_font_size="8pt"
        )
    )

    overlays.append(
        hv.Labels(phospho_labels).opts(
            text_font_size="11pt",
            text_font_style="bold"
        )
    )

    return hv.Overlay(overlays)
```

```python
# Combine b-ions and y-ions into a single list of theoretical ions for matching
theoretical_ions = b_ions + y_ions

# Match the theoretical ions to the observed spectrum using the has_peak method with a ppm tolerance of 10 ppm
matches = match_ions_haspeak(theoretical_ions, spec, ppm_tol=10)

# Create an annotated plot of the spectrum with matched peaks highlighted
annotated_plot = highlight_matches(ms_spikes, matches, ppm=10)

annotated_plot
```
