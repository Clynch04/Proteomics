```python
import pymzml
from pyteomics import parser
from Bio import SeqIO
import pandas as pd
import os
import sys
import holoviews as hv
hv.extension('bokeh')
```

```python
# Find a protein with at least 5 tryptic peptides
for record in SeqIO.parse("methanococcus_maripaludis_proteome.fasta", "fasta"):
    protein_id = str(record.id)
    sequence = str(record.seq)
    tryptic_peptides = parser.cleave(sequence, "trypsin")
    
    if len(tryptic_peptides) >= 5:
        # Store for later use
        selected_protein_id = protein_id
        selected_peptides = list(tryptic_peptides)
        
        print(f"Protein ID: {selected_protein_id}")
        print(f"Total peptides: {len(tryptic_peptides)}")
        for i, peptide in enumerate(selected_peptides, 1):
            print(f"{i}. {peptide}")
        break
```

```python
from pyteomics import mass

# Amino acid monoisotopic masses
def calculate_peptide_mz(peptide_sequence, charge_state):
    """Calculate m/z for a peptide at a given charge state"""
    # Get monoisotopic mass
    mono_mass = mass.calculate_mass(sequence=peptide_sequence, ion_type='M')
    
    # Calculate m/z
    mz = (mono_mass + charge_state * 1.0073) / charge_state
    return mz

# Create a theoretical peptide database
theoretical_peptides = []

for peptide in selected_peptides:
    mono_mass = mass.calculate_mass(sequence=peptide, ion_type='M')
        
    for charge in [2, 3]:
        mz = calculate_peptide_mz(peptide, charge)
        theoretical_peptides.append({
           'peptide': peptide,
           'length': len(peptide),
            'monoisotopic_mass': mono_mass,
            'charge': charge,
            'mz': mz,
        })

# Convert to DataFrame for easy analysis
df_peptides = pd.DataFrame(theoretical_peptides)
print(f"Generated {len(df_peptides)} theoretical peptide ions")
print(df_peptides)
```

```python
# Load the mzML file
mzml_file = "2020-08-16_I3_0_2.mzML"
run = pymzml.run.Reader(mzml_file)
```

```python
# Define 10 ppm tolerance function
def ppm_tolerance(theoretical_mz, ppm=10):
    """Calculate m/z tolerance in Daltons for a given ppm"""
    return (theoretical_mz * ppm) / 1e6

# Store matched intensities for each peptide
peptide_intensities = {
    idx: {'peptide': row['peptide'], 'charge': row['charge'], 'mz': row['mz'], 'intensities': []}
    for idx, row in df_peptides.iterrows()
}

# Iterate through MS1 scans
for spec in run:
    if spec['ms level'] == 1:
        # Get m/z and intensity arrays from spectrum
        mz_array = spec.mz
        intensity_array = spec.i
        
        # For each theoretical peptide, find matching peaks
        for idx, row in df_peptides.iterrows():
            theoretical_mz = row['mz']
            tolerance = ppm_tolerance(theoretical_mz, ppm=10)
            
            # Find matching peaks in this spectrum
            for obs_mz, obs_intensity in zip(mz_array, intensity_array):
                if abs(obs_mz - theoretical_mz) <= tolerance:
                    peptide_intensities[idx]['intensities'].append({
                        'scan': spec.ID,
                        'retention_time': spec.scan_time_in_minutes(),
                        'observed_mz': obs_mz,
                        'intensity': obs_intensity
                    })

# Display results
print(f"Matched peptides across {len(list(run))} scans:\n")
for idx, data in peptide_intensities.items():
    print(f"Peptide: {data['peptide']} (Charge: +{data['charge']}, m/z: {data['mz']:.4f})")
    print(f"  Found in {len(data['intensities'])} spectra")
    if data['intensities'][:3]:  # Show first 3 matches
        for match in data['intensities'][:3]:
            print(f"    - Scan{match['scan']}: m/z={match['observed_mz']:.4f}, intensity={match['intensity']:.0f}")
    print()
```

```python
# First, get all MS2 scan parent ions and their RTs
ms2_parents = {}
for spec in pymzml.run.Reader(mzml_file):
    if spec['ms level'] == 2:
        # Get precursor information
        precursors = spec.selected_precursors
    
        if precursors:  # Check if precursor exists
            for precursor in precursors:
                precursor_mz = precursor['mz']
                precursor_charge = precursor['charge']
        
        rt = spec.scan_time_in_minutes()
        
        if (precursor_mz, precursor_charge) not in ms2_parents:
            ms2_parents[(precursor_mz, precursor_charge)] = []
        ms2_parents[(precursor_mz, precursor_charge)].append(rt)

# Create holoviews plots for each peptide
plots = []

for idx, row in df_peptides.iterrows():
    data = peptide_intensities[idx]
    
    # Sort by retention time
    sorted_intensities = sorted(data['intensities'], key=lambda x: x['retention_time'])
    
    if sorted_intensities:
        rts = [m['retention_time'] for m in sorted_intensities]
        intensities = [m['intensity'] for m in sorted_intensities]
        
        # Create DataFrame for curve
        eic_data = pd.DataFrame({'RT': rts, 'Intensity': intensities})
        
        # Create the EIC curve
        eic_curve = hv.Curve(eic_data, 'RT', 'Intensity', label='EIC')
        
        # Find and mark MS2 selection points
        theoretical_mz = row['mz']
        charge = row['charge']
        tolerance = ppm_tolerance(theoretical_mz, ppm=10)
        
        ms2_rts = []
        for (parent_mz, parent_charge), rt_list in ms2_parents.items():
            if parent_charge == charge and abs(parent_mz - theoretical_mz) <= tolerance:
                ms2_rts.extend(rt_list)
        
        # Create scatter plot for MS2 points
        if ms2_rts:
            # Get intensities at MS2 RTs
            ms2_intensities = []
            for ms2_rt in ms2_rts:
                # Find closest point in EIC
                closest_idx = min(range(len(rts)), key=lambda i: abs(rts[i] - ms2_rt))
                ms2_intensities.append(intensities[closest_idx])
            
            ms2_data = pd.DataFrame({'RT': ms2_rts, 'Intensity': ms2_intensities})
            ms2_scatter = hv.Scatter(ms2_data, 'RT', 'Intensity', label='MS2 Selected').opts(color='red', size=10)
            plot = (eic_curve * ms2_scatter)
        else:
            plot = eic_curve
        
        # Format the plot
        plot = plot.opts(
            title=f'{data["peptide"]} (Charge: +{data["charge"]}, m/z: {data["mz"]:.4f})',
            xlabel='Retention Time (minutes)',
            ylabel='Intensity',
            width=900,
            height=300,
            ylim=(0, None)
        )
        
        plots.append(plot)

# Combine all plots
if plots:
    eic_viz = hv.Layout(plots).cols(1)
    display(eic_viz)
```
