#### This Jupyter Notebook demonstrates how to read an mzML file, extract retention time and total ion current data from MS1 spectra, and visualize the chromatogram using Plotly.

### Step 1: Importing Required Libraries

We need the following Python libraries:
- `pymzml` to read mzML mass spectrometry files.
- `plotly.graph_objects` to create interactive plots.

be sure to 
`pip install`
plotly and pymzml


```python
import pymzml
import pandas as pd
import os
import sys
import holoviews as hv
hv.extension('bokeh')
```

### Step 2: Reading the mzML File

We will load an mzML file and extract retention time and TIC values.

```python
mzml_file = "2020-08-16_I3_0_2.mzML" #replace with your mzML file
run = pymzml.run.Reader(mzml_file)

# Iterate through spectra and extract spectrum IDs and retention times
for n, spec in enumerate(run):
    last_spec = spec.ID
    last_rt = spec.scan_time_in_minutes()
print(last_spec)
print(last_rt)
```

### Step 3: MS1 vs MS2 spectra

```python
rts = []
ms_levels = []
colors = []

# Open the file
run = pymzml.run.Reader(mzml_file)

for spec in run:
    rt = spec.scan_time_in_minutes()
    
    # Extract MS level
    ms_lvl = spec.ms_level
    
    rts.append(rt)
    ms_levels.append(ms_lvl)
    
    # Assign color based on MS level
    if ms_lvl == 1:
        colors.append('blue')
    else:
        colors.append('red')

print(f"Parsed {len(rts)} spectra.")

# Create a DataFrame for HoloViews
df_scans = pd.DataFrame({
    'Retention Time': rts,
    'MS Level': ms_levels,
    'Color': colors
})

# Create the HoloViews Scatter plot
scatter = hv.Scatter(
    df_scans, 
    kdims=['Retention Time'], 
    vdims=['MS Level', 'Color']
).opts(
    # Visual options
    color='Color',          # Use the 'Color' column for point colors
    size=5,
    alpha=0.6,
    width=800,
    height=400,
    
    # Axes and Title
    xlabel="Retention Time (min)",
    ylabel="MS Level",
    title="MS Scan Distribution by Retention Time",
    yticks=[1, 2],          # Only show ticks at 1 and 2
    ylim=(0.5, 2.5),        # Add padding
    
    # Tools
    tools=['hover', 'box_zoom', 'reset']
)

scatter
```

### Step 4: Plotting a spectrum

Here we will visualize an MS1 spectrum from the middle of the MS run

```python
import pandas as pd
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')

spec = run[20016] # select a spectrum by its index
print("Spectrum MS level:", spec.ms_level)

mz_list = []
intensities_list = []
for mz, i in spec.peaks("centroided"):
    mz_list.append(mz)
    intensities_list.append(i)

# Put the mz values and intensities into a pandas dataframe
df_spec = pd.DataFrame({ "mz": mz_list, "intensity": intensities_list })

# Create the plot
ms1_spikes = hv.Spikes(
    df_spec, 
    kdims="mz", 
    vdims="intensity"
).opts(
    width=900,
    height=350,
    line_width=1.0,
    color='black',
    tools=['box_zoom', 'hover', 'pan', 'wheel_zoom', 'reset', 'save'],
    active_tools=['box_zoom'],
    ylim=(0, None),
    xlabel="m/z",
    ylabel="Intensity",
    title="MS1 Spectrum"
)

ms1_spikes
```

### Step 5: Finding a peak in a spectrum

```python
spec = run[20016]
target_mz = 794.92
peak_list = spec.has_peak(target_mz)

if peak_list:
    mz_found = peak_list[0][0]
    intensity_found = peak_list[0][1]
    
    # 2. Calculate the width (Delta) based on 5 ppm
    # Delta = (ppm * mz) / 1e6
    ppm = 5
    delta_mz = (ppm * target_mz) / 1e6
    
    # 3. Define the 3 points of the triangle
    # (Left Base, Peak, Right Base)
    triangle_data = pd.DataFrame({
        'mz': [target_mz - delta_mz, target_mz, target_mz + delta_mz],
        'intensity': [0, intensity_found, 0]
    })
    
    # 4. Create the Triangle using hv.Area or hv.Curve
    # hv.Area fills it in, which looks nice for a "peak shape"
    peak_triangle = hv.Area(
        triangle_data, kdims='mz', vdims='intensity'
    ).opts(
        color='red',
        alpha=0.5,       # Semi-transparent so you can see the original spike behind it
        line_width=1,
        line_color='red'
    )
    
    # 5. Overlay with the original spectrum
    # (Assuming ms1_spikes is already defined from previous steps)
    overlay = (ms1_spikes * peak_triangle).opts(
        width=900, 
        height=350,
        title=f"MS1 Spectrum with Match (Error: +/- {ppm} ppm)",
        xlabel="m/z", 
        ylabel="Intensity",
        active_tools=['box_zoom']
    )
    
    display(overlay)

else:
    print(f"No peak found near {target_mz}")
```
