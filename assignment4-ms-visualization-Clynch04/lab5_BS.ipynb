```python
import pymzml
from pyteomics import parser
from Bio import SeqIO
import pandas as pd
import os
import sys
import holoviews as hv
hv.extension('bokeh')
```

```python
pep_dict = {}
for record in SeqIO.parse("methanococcus_maripaludis_proteome.fasta", "fasta"):
    protein_id = str(record.id)
    sequence = str(record.seq)
    # use pyteomics to generate tryptic peptides
    tryptic_peptides = parser.cleave(sequence, "trypsin")
    for pep in tryptic_peptides:
        if pep in pep_dict:
            pep_dict[pep].append(protein_id)
        else:
            pep_dict[pep] = [protein_id]
```

```python
from pyteomics import mass

# Amino acid monoisotopic masses
def calculate_peptide_mz(peptide_sequence, charge_state):
    """Calculate m/z for a peptide at a given charge state"""
    # Get monoisotopic mass
    mono_mass = mass.calculate_mass(sequence=peptide_sequence, ion_type='M')
    
    # Calculate m/z
    mz = (mono_mass + charge_state * 1.0073) / charge_state
    return mz

# Create a theoretical peptide database
theoretical_peptides = []

for peptide, proteins in pep_dict.items():
    # Filter by length (5-40)
    if 5 <= len(peptide) <= 40:
        mono_mass = mass.calculate_mass(sequence=peptide, ion_type='M')
        
        for charge in [2, 3]:
            mz = calculate_peptide_mz(peptide, charge)
            theoretical_peptides.append({
                'peptide': peptide,
                'length': len(peptide),
                'monoisotopic_mass': mono_mass,
                'charge': charge,
                'mz': mz,
                'proteins': proteins
            })

# Convert to DataFrame for easy analysis
df_peptides = pd.DataFrame(theoretical_peptides)
print(f"Generated {len(df_peptides)} theoretical peptide ions")
print(df_peptides.head())
```

```python
# Code cell taken from starter code

mzml_file = "2020-08-16_I3_0_2.mzML"
run = pymzml.run.Reader(mzml_file)

spec = run[20016] # select a spectrum by its index
print("Spectrum MS level:", spec.ms_level)

mz_list = []
intensities_list = []
for mz, i in spec.peaks("centroided"):
    mz_list.append(mz)
    intensities_list.append(i)

# Put the mz values and intensities into a pandas dataframe
df_spec = pd.DataFrame({ "mz": mz_list, "intensity": intensities_list })

# Create the plot
ms1_spikes = hv.Spikes(
    df_spec, 
    kdims="mz", 
    vdims="intensity"
).opts(
    width=900,
    height=350,
    line_width=1.0,
    color='black',
    tools=['box_zoom', 'hover', 'pan', 'wheel_zoom', 'reset', 'save'],
    active_tools=['box_zoom'],
    ylim=(0, None),
    xlabel="m/z",
    ylabel="Intensity",
    title="MS1 Spectrum"
)

ms1_spikes
```

```python
# Match theoretical peptide m/z values against experimental MS1 peaks

ppm_tolerance = 10

# Get peaks from the spectrum as a list
peaks_in_spectrum = df_spec['mz'].values

# Find matches
matches = []
for idx, row in df_peptides.iterrows():
    theoretical_mz = row['mz']
    
    # Calculate ppm tolerance window
    delta_mz = (ppm_tolerance * theoretical_mz) / 1e6
    lower_bound = theoretical_mz - delta_mz
    upper_bound = theoretical_mz + delta_mz
    
    # Check if any peak falls within the tolerance window
    matching_peaks = peaks_in_spectrum[(peaks_in_spectrum >= lower_bound) & 
                                       (peaks_in_spectrum <= upper_bound)]
    
    if len(matching_peaks) > 0:
        matches.append({
            'peptide': row['peptide'],
            'charge': row['charge'],
            'theoretical_mz': theoretical_mz,
            'observed_mz': matching_peaks[0],  # Use first match if multiple
            'ppm_error': ((matching_peaks[0] - theoretical_mz) / theoretical_mz) * 1e6,
            'proteins': row['proteins']
        })

# Convert matches to DataFrame
df_matches = pd.DataFrame(matches)

print(f"Total peptide ions in theoretical database: {len(df_peptides)}")
print(f"Peptide ions matching peaks (10 ppm tolerance): {len(df_matches)}")
print(f"Match percentage: {(len(df_matches)/len(df_peptides)*100):.1f}%")
print("\nTop 10 matches:")
print(df_matches.head(10))
```

```python
# Code cell taken from starter code with different target m/z and ppm tolerance

spec = run[20016]
target_mz = 516.27
peak_list = spec.has_peak(target_mz)

if peak_list:
    mz_found = peak_list[0][0]
    intensity_found = peak_list[0][1]
    
    # 2. Calculate the width (Delta) based on 5 ppm
    # Delta = (ppm * mz) / 1e6
    ppm = 10
    delta_mz = (ppm * target_mz) / 1e6
    
    # 3. Define the 3 points of the triangle
    # (Left Base, Peak, Right Base)
    triangle_data = pd.DataFrame({
        'mz': [target_mz - delta_mz, target_mz, target_mz + delta_mz],
        'intensity': [0, intensity_found, 0]
    })
    
    # 4. Create the Triangle using hv.Area or hv.Curve
    # hv.Area fills it in, which looks nice for a "peak shape"
    peak_triangle = hv.Area(
        triangle_data, kdims='mz', vdims='intensity'
    ).opts(
        color='red',
        alpha=0.5,       # Semi-transparent so you can see the original spike behind it
        line_width=1,
        line_color='red'
    )
    
    # 5. Overlay with the original spectrum
    # (Assuming ms1_spikes is already defined from previous steps)
    overlay = (ms1_spikes * peak_triangle).opts(
        width=900, 
        height=350,
        title=f"MS1 Spectrum with Match (Error: +/- {ppm} ppm) | target m/z: {target_mz}",
        xlabel="m/z", 
        ylabel="Intensity",
        active_tools=['box_zoom']
    )
    
    display(overlay)

else:
    print(f"No peak found near {target_mz}")
```

```python
# Match MS2 precursor ions to theoretical peptides
ppm_tolerance = 10

ms2_start_idx = 20017
ms2_end_idx = 20031

ms2_matches = []

# Iterate through MS2 spectra
for spec_idx in range(ms2_start_idx, ms2_end_idx + 1):
    spec = run[spec_idx]
    
    # Get precursor information
    precursors = spec.selected_precursors
    
    if precursors:  # Check if precursor exists
        for precursor in precursors:
            precursor_mz = precursor['mz']
            precursor_charge = precursor['charge']
            
            # Calculate ppm tolerance window
            delta_mz = (ppm_tolerance * precursor_mz) / 1e6
            lower_bound = precursor_mz - delta_mz
            upper_bound = precursor_mz + delta_mz
            
            # Find matching theoretical peptides (both m/z and charge must match)
            matching_peptides = df_peptides[
                (df_peptides['mz'] >= lower_bound) & 
                (df_peptides['mz'] <= upper_bound) &
                (df_peptides['charge'] == precursor_charge)
            ]
            
            if len(matching_peptides) > 0:
                for idx, pep in matching_peptides.iterrows():
                    ms2_matches.append({
                        'spectrum_index': spec_idx,
                        'precursor_mz': precursor_mz,
                        'precursor_charge': precursor_charge,
                        'peptide': pep['peptide'],
                        'theoretical_mz': pep['mz'],
                        'ppm_error': ((precursor_mz - pep['mz']) / pep['mz']) * 1e6,
                        'proteins': pep['proteins']
                    })

# Convert to DataFrame
df_ms2_matches = pd.DataFrame(ms2_matches)

# Count unique MS2 spectra with matches
unique_ms2_spectra = df_ms2_matches['spectrum_index'].nunique() if len(df_ms2_matches) > 0 else 0

print(f"Total MS2 spectra analyzed: {ms2_end_idx - ms2_start_idx + 1} (indices {ms2_start_idx}-{ms2_end_idx})")
print(f"MS2 spectra with matching precursor peptides: {unique_ms2_spectra}")
print(f"Total precursor-peptide matches: {len(df_ms2_matches)}")

if len(df_ms2_matches) > 0:
    print(f"\nMatch percentage: {(unique_ms2_spectra/(ms2_end_idx - ms2_start_idx + 1)*100):.1f}% of MS2 spectra")
    print("\nAll matches:")
    print(df_ms2_matches[['spectrum_index', 'precursor_mz', 'precursor_charge', 'peptide', 'ppm_error']])
else:
    print("No matching peptides found")
```
